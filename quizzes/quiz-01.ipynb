## 3. (2 pts) Iterate and print

Iterate over `months` and print `"my birthday"` when the month is your birth
month.

````python
months = ["January","February","March","April","May","June","July","August","September","October","November","December"]

my_birthday = "December"
i = 0
while i < len(months):
    if months[i] == my_birthday:
        print("My birthday is in " + months[i])
        break
    i += 1
````

## 4. (3 pts) Variables, objects, literals, expressions

What is the relationship between them? Give a code example.

Variables reference objects. Objects are the actual thing that exists in memory that has properties. Literal is a value that is written directly to source code. An expression is anything that evaluates to a value. A literal is a kind of expression, evaluating an expression produces an object, and a variable is then bound to that object

````python
x = 5
y = x
print(type(x))
print(type(y))
print(id(x))
print(id(y))
print(x is y)
````

## 5. (2 pts) Immutable sequences

What are they? Give a code example showing that one is immutable.

str, tuple, bytes, range are the immutable sequences. list and bytearray are the mutable ones.

````python
T = (1, 2, 3, 4, 5)
len(T)

T + (6, 7, 8, 9, 10)

try:
    T[0] = 10
except TypeError as e:
    print(f"Error: {e}")

T = (10,) + T[1:]
T
````
