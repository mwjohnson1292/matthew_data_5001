1. (3 pts) The string "a\nb\x1f\000d" contains the bytes a, newline (\n), b, binary 31 (a
hex escape \x1f), binary 0 (an octal escape \000), and d. 
- Pass the string to the built-
in len function to verify this, and print each of its character’s ord results to see the
actual byte values. See Table 7-2 for more details.
- Convert the string to a human readable format

```python
# Pass the string to the built-in len function to verify this, and print each of its character’s ord results to see theactual byte values.

s = "a\nb\x1f\000d"
print(f"Length: {len(s)} characters")

print("")
for ch in s:
    print(f"{ch!r:<8} {ord(ch):>4}")

# convert the string to a human readable format
print(f"\nHuman-readable: {s!r}")
```

2. Dictionary Practice (10 pts)

data = {'state':['Ohio','Ohio','Ohio','Nevada','Nevada','Nevada'],

'year':[2000,2001,2002,2001,2002,2003],

'pop':[1.5, 1.7, 3.6, 2.4, 2.9, 3.2]}

- Use a Dictionary method to print how many keys are in data.
- Use an if/else statement to validate that the 'pop' key exists in the data object. If the "pop" key does not exist, print "key not found".
- Update the 'year' key's list and append an item of your choice.
- Use the pop method to remove the state key.
- Create a new Dictionary from the 'state' key and sort it alphabetically.
- Loop through data and print the key name and key length
- Access the 'pop' key and slice the list to exclude the first element.
- Add a new dictionary key called 'region' and set it to ['Midwest', 'Midwest', 'Midwest','West','West','West']

```python
data = {'state':['Ohio','Ohio','Ohio','Nevada','Nevada','Nevada'],
'year':[2000,2001,2002,2001,2002,2003],
'pop':[1.5, 1.7, 3.6, 2.4, 2.9, 3.2]}

# use dictionary method to print how many keys are in data
print(f"Number of keys in data: {len(data.keys())}")

# - Use an if/else statement to validate that the 'pop' key exists in the data object. If the "pop" key does not exist, print "key not found".
if 'pop' in data:
    print(f"The 'pop' key exists in the data object.")
else:
    print(f"Key not found.")

#- Update the 'year' key's list and append an item of your choice.
data['year'].append(2004)
print(f"The updated data object is: {data}")

#- Use the pop method to remove the state key.
removed_states = data.pop('state')
print(f"The updated data object is: {data}")

#- Create a new Dictionary from the 'state' key and sort it alphabetically.
state_dict = {'state': sorted(removed_states)}
print(f"The sorted state dictionary is: {state_dict}")

#- Loop through data and print the key name and key length
for key in data:
    print(f"Key: {key}, Length: {len(data[key])}")

#- Access the 'pop' key and slice the list to exclude the first element.
pop_values = data['pop'][1:]
print(f"The 'pop' key values (excluding the first element) are: {pop_values}")

#- Add a new dictionary key called 'region' and set it to ['Midwest', 'Midwest', 'Midwest','West','West','West']
data['region'] = ['Midwest', 'Midwest', 'Midwest','West','West','West']
print(f"The updated data object with 'region' key is: {data}")
```

```python
states = [' Alabama ', 'Georgia!', 'Georgia', 'georgia','FlOrIda', 'south carolina##', 'West virginia?']

normalized = []
for s in states:
    normalized.append(s.strip(' !#?').title())

print(f"The normalized states list is: {normalized}")
```

6. (2 pts) Loop through the string and eliminate all the asterisks.

```python
asterisk = 'My* compu*ter has go*ne craz***y an*d add*ed extra *symbols.*'

cleaned = ''
for char in asterisk:
    if char != '*':
        cleaned += char
print(f"The cleaned string is: {cleaned}")
```

7. (2 pts) Iterate through integers_list and print values that are multiples of 9.
