## Question 1 (2 pts)

*Raise a `ZeroDivisionError` exception using the `raise`
statement. Does the `else` execute? Include your findings as a comment in your program.*

````python
numerator = 3.14
denominator = 0

try:
    if denominator == 0:
        raise ZeroDivisionError("denominator is zero")
    result = numerator / denominator
except ZeroDivisionError as e:
    print("except ran:", e)
else:
    print("else ran:", result)

# Findings: else does NOT run. raise jumps straight to except;
# else only runs when the try block raises nothing.
````

## Question 3 (2 pts)

*Why is `except Exception as e:` recommended over the generic `except`?*

- A bare `except:` catches everything, including Ctrl+C (`KeyboardInterrupt`) and `SystemExit`, which can leave a program you can't stop. `except Exception` catches only real errors.
- `as e` keeps the error object, so you can see what went wrong instead of hiding it.
