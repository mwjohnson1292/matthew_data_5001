8. (2 pts) What is the difference between a dynamically typed and statically typed language? 

Which one is python? 

When is that a good thing? 

When is that a bad thing? 

```python
x = 1
print(type(x))

x = "now i am a string"
print(type(x))

# Python is a dynamically typed language.

# Dynamically typed means that the type of a variable is determined at runtime and can change as the program executes.
# Statically typed languages require the type of a variable to be declared at compile time and do not allow it to change at runtime.

# Dynamically typed is good when you want flexibility and don't want to explicitly declare variable types.
# Dynamically typed can be bad because it may lead to runtime errors that would have been caught at compile time in a statically typed language.
```

20. (5 pts) Using the file you just imported: 

For each line of the file:
Use the find() function to get the index of "STL_" in the line.
Slice the string so that only the characters that follow "STL_" are printed.  Example: STL_Clayton would print Clayton. 

Each line has a value with a standard prefix of "STL_". We want to remove the "STL_" prefix using slicing. 

```python
with open("StLouisMunicipalities.csv") as f:
    for line in f:
        line = line.strip()
        clean = line[line.find("STL_") + len("STL_"):]
        print(clean)

# This code reads a CSV file containing St. Louis municipalities, extracts the part of each line after "STL_", and prints it.
```
