4. (3 pts) Create a new pandas dataframe that has the average mpg, weight, and horsepower per country of origin stored as a tuple in that column. What does this tell us? Why might you use a tuple? Be mindful of how you name the columns so information isnt lost. 

```python
import pandas as pd

df = pd.read_csv("cars.csv", sep=";")
df_headers = df.columns
print(df_headers)

# check all car's metrics for valid values
df_real = df[df["MPG"] > 0]
df_real = df_real[df_real["Cylinders"] > 0]
df_real = df_real[df_real["Displacement"] > 0]
df_real = df_real[df_real["Horsepower"] > 0]
df_real = df_real[df_real["Weight"] > 0]
df_real = df_real[df_real["Acceleration"] > 0]
print(f"\nbefore cleaning: {len(df)} rows")
print(f"after cleaning: {len(df_real)} rows")

# calculate average mpg, weight, and horsepower by origin
avg_mpg_by_origin = df_real.groupby("Origin")["MPG"].mean().round(2)
avg_weight_by_origin = df_real.groupby("Origin")["Weight"].mean().round(2)
avg_horsepower_by_origin = df_real.groupby("Origin")["Horsepower"].mean().round(2)

# build a 2 column dataframe with origin and tuple of average metrics
europe = (avg_mpg_by_origin["Europe"], avg_weight_by_origin["Europe"], avg_horsepower_by_origin["Europe"])
us = (avg_mpg_by_origin["US"], avg_weight_by_origin["US"], avg_horsepower_by_origin["US"])
japan = (avg_mpg_by_origin["Japan"], avg_weight_by_origin["Japan"], avg_horsepower_by_origin["Japan"])

df_avg_metrics = pd.DataFrame({
    "Origin": ["Europe", "US", "Japan"],
    "(Average mpg, weight, horsepower)": [europe, us, japan]
})
print(f"\naverage metrics by origin:\n{df_avg_metrics}")
```

Reflection: *This totally went against all my instincts which was kind of fun to work through. At work, I always have to find ways to split or parse concatenated columns but I have never actually had to build one from scratch like this. Aside from the concatenation making data analysis or data science more difficult, using tuples is also challenging in that the position matters and there is no way to reference labels. The advantage is that when it is stored as a bundled result, it can't be changed because immutability offers integrity. But we can see that US cars are much heavier and more powerful but get worse gas mileage compared to Europe and Japan.*

Identify the indicies of the books with a status of in-library, and print the corresponding book titles.

```python
# Part 1 - a list of dictionaries: one dictionary per book
books = [
    {"title": "The Art of War", "author": "Sun Tzu", "subject": "strategy", "year": -500, "status": "on-loan"},
    {"title": "Meditations", "author": "Marcus Aurelius", "subject": "stoicism", "year": 180, "status": "in-library"},
    {"title": "Tao Te Ching", "author": "Lao Tzu", "subject": "taoism", "year": -400, "status": "in-library"},
    {"title": "Shaolin Spirit", "author": "Shi Heng Yi", "subject": "self-mastery", "year": 2023, "status": "in-library"},
    {"title": "A Brief History of Time", "author": "Stephen Hawking", "subject": "cosmology", "year": 1988, "status": "on-loan"},
    {"title": "Cosmos", "author": "Carl Sagan", "subject": "astronomy", "year": 1980, "status": "in-library"},
    {"title": "The Elegant Universe", "author": "Brian Greene", "subject": "string theory", "year": 1999, "status": "on-loan"},
    {"title": "Parallel Worlds", "author": "Michio Kaku", "subject": "physics", "year": 2004, "status": "on-loan"},
]

# Part 3 - access each book's components by indexing
print("Accessing each book's components by indexing:")
for i in range(len(books)):
    print(f"i: {i}, title: {books[i]['title']}, author: {books[i]['author']}, subject: {books[i]['subject']}, year: {books[i]['year']}, status: {books[i]['status']}")

# Part 4 - write the book collection to a csv file
import csv

with open("books.csv", "w", newline="") as csvfile:
    fieldnames = ["title", "author", "subject", "year", "status"]
    writer = csv.DictWriter(csvfile, fieldnames=fieldnames)

    writer.writeheader()
    for book in books:
        writer.writerow(book)

# Part 5 - read the csv back and build one list per component
print("\nReading the csv back and building one list per component:")
titles = []
authors = []
subjects = []
years = []
statuses = []

with open("books.csv", "r") as csvfile:
    reader = csv.DictReader(csvfile)
    for row in reader:
        titles.append(row["title"])
        authors.append(row["author"])
        subjects.append(row["subject"])
        years.append(int(row["year"])) # need to cast to int, otherwise, just text
        statuses.append(row["status"])

print(f"titles: {titles}")
print(f"authors: {authors}")
print(f"subjects: {subjects}")
print(f"years: {years}")
print(f"statuses: {statuses}")

# Part 6 - find the indicies of the in-library books, then print their titles
print("\nFinding the indicies of the in-library books, then printing their titles:")
in_library_indices = []
for i in range(len(statuses)):
    if statuses[i] == "in-library":
        in_library_indices.append(i)

for i in in_library_indices:
    print(f"index: {i}, title: {titles[i]}")
```
